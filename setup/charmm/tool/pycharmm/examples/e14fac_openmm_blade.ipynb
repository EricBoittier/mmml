## Generate nbfix files for aritmatic and geometric combining rules

```python
%%script false --no-raise-error
# magic command that disables execution of cell
def retreive_nbData(files=[]):
    import numpy as np
    nb = {}
    nbfix = {}
    for name in files:
        key = name.split('/')[-1].split('.')[0]
        print('Reading file', key)
        params = open(name,'r').readlines()
        read_nbond = False
        read_nbfix = False
        for line in params:
            if len(line) == 0: continue
            if line[0:1] == '*': continue
            if line[0:4].upper() == 'NONB': 
                read_nbond = True
            elif line[0:4].upper() == 'NBFI': 
                read_nbfix = True
                read_nbond = False
            elif read_nbond:
                cl = line.split('!')[0].split()
                if len(cl) == 4:
                    nb[cl[0]] = [float(cl[2]),float(cl[3])]
                elif len(cl) == 7:
                    nb[cl[0]] = [float(cl[2]),float(cl[3]),float(cl[5]),float(cl[6])]
            elif read_nbfix:
                cl = line.split('!')[0].split()
                if len(cl) == 4:
                    nbfix[f'{cl[0]}-{cl[1]}'] = [float(cl[2]),float(cl[3])]
                elif len(cl) == 6:
                    nbfix[f'{cl[0]}-{cl[1]}'] = [float(cl[2]),float(cl[3]),float(cl[4]),float(cl[5])]
            elif line[0:4].upper() == 'HBON': break
            elif line[0:3].upper() == 'END':
                read_nbfix = False
                read_nbond = False
    return (nb,nbfix)

def generate_nbfix(nbD1,nbfixD1,nbD2, nbfixD2, defD1 = None,
                   name=None,arith=False,geom=False):
    import numpy as np
    same = nbD1 == nbD2 and nbfixD1 == nbfixD2
    print(f'Dictionaries are the same={same}')
    nbfix_new = {}
    # Combine into master nbfix dictionary
    def combine(a,b,geom):
        if geom:
            return np.sqrt(a*b)*2
        return a+b
    for i, k1 in enumerate(nbD1.keys()):
        for j, k2 in enumerate(nbD2.keys()):
            if same and j<=i: continue
            nk = f'{k1}-{k2}'
            nkr = f'{k2}-{k1}'
            if same and nkr in nbfixD1: nk = nkr
            # If processing self-self interactions make sure
            # original nbfixes are retained if processing is with same
            # combination-type
            if same and (nk in nbfixD1) and ((defD1 == 'arith' and arith)\
                or (defD1 == 'geom' and geom)):
                    #print(nk,nbfixD1.keys(),name)
                    nbfix_new[nk] = nbfixD1[nk]
                    continue
            nbfix_new[nk] = [-np.sqrt(nbD1[k1][0]*nbD2[k2][0]),
                             combine(nbD1[k1][1],nbD2[k2][1],geom)]
            if len(nbD1[k1])==2 and len(nbD2[k2])==4:
                nbfix_new[nk].append(-np.sqrt(nbD1[k1][0]*nbD2[k2][2]))
                nbfix_new[nk].append(combine(nbD1[k1][1],nbD2[k2][3],geom))
            elif len(nbD1[k1])==4 and len(nbD2[k2])==2:
                nbfix_new[nk].append(-np.sqrt(nbD1[k1][2]*nbD2[k2][0]))
                nbfix_new[nk].append(combine(nbD1[k1][3],nbD2[k2][1],geom))
            elif len(nbD1[k1])==4 and len(nbD2[k2])==4:
                nbfix_new[nk].append(-np.sqrt(nbD1[k1][2]*nbD2[k2][2]))
                nbfix_new[nk].append(combine(nbD1[k1][3],nbD2[k2][3],geom))

    if arith: pout = open(f'lig_test/nbfix_{name}_arith.prm','w')
    if geom: pout = open(f'lig_test/nbfix_{name}_geom.prm','w')
    print(f'* NBFIX Table for atoms\n* {name}\n*   \nNBFIX',file=pout)
    for k in nbfix_new.keys():
        line = f'{k.split("-")[0]}  {k.split("-")[1]}  '
        for v in nbfix_new[k]:
            line +=f'{v:.5f}  '
        print(line,file=pout)
    print('END',file=pout)
    pout.close()
    return
              
nb_prot,nbfix_prot = retreive_nbData(['/home/brookscl/charmm/charmm-latest/toppar/par_all36m_prot.prm',
                '/home/brookscl/charmm/charmm-latest/toppar/toppar_water_ions.str'])
nb_lig,nbfix_lig = retreive_nbData(['lig_test/UNK_5E51CD.prm'])
generate_nbfix(nb_prot,nbfix_prot,nb_prot,nbfix_prot,'arith',name='c36',
               arith=True)
generate_nbfix(nb_prot,nbfix_prot,nb_prot,nbfix_prot,'arith',name='c36',
               geom=True)
generate_nbfix(nb_lig,nbfix_lig,nb_lig,nbfix_lig,'geom',name='ligand',
               arith=True)
generate_nbfix(nb_lig,nbfix_lig,nb_lig,nbfix_lig,'geom',name='ligand',
               geom=True)
generate_nbfix(nb_prot,nbfix_prot,nb_lig,nbfix_lig,'arith',name='c36-ligand',
               arith=True)
generate_nbfix(nb_prot,nbfix_prot,nb_lig,nbfix_lig,'arith',name='c36-ligand',
               geom=True)
```
